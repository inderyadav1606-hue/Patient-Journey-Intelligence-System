### 1. Load Enriched Patient Dataset


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath('../src'))
from clustering import PatientSegmentationEngine

df = pd.read_csv('../data/processed_patient_master.csv')
df.head()
```

### 2. K-Means Clustering & 2D Latent Space Projection


```python
segmenter = PatientSegmentationEngine(n_clusters=4, model_dir='../models')
df_clustered, sil_score, cluster_means = segmenter.fit_segmentation(df)

print(f"Optimal Cluster Silhouette Score: {sil_score:.3f}")
print("Cluster Membership Distribution:\n", df_clustered['Cohort_Name'].value_counts())
```

### 3. PCA Latent Space Visualization


```python
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df_clustered,
    x='PCA_1', y='PCA_2',
    hue='Cohort_Name',
    palette='Set1',
    alpha=0.85,
    s=60
)
plt.title("Patient Phenotyping Clusters (PCA Dimensionality Reduction)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

### 4. Cohort Biomarker Profiles


```python
summary = segmenter.get_cluster_profiles(df_clustered)
summary[['Cohort_Name', 'Age', 'Acute_Utilization_Ratio', 'Avg_Systolic_BP', 'Avg_Glucose', 'Avg_eGFR', 'Target_Risk']]
```
