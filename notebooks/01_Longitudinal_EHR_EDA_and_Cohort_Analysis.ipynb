### 1. Ingestion & Module Imports


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to path
sys.path.append(os.path.abspath('../src'))
from data_pipeline import HealthcareDataPipeline

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)
```

### 2. Run Data Cleaning & Longitudinal Feature Extraction


```python
pipeline = HealthcareDataPipeline(data_dir='../data')
master_df = pipeline.run_full_pipeline()
master_df.head()
```

### 3. Patient Demographics & Condition Prevalence


```python
print("Dataset shape:", master_df.shape)
print("Target Risk Class Balance:\n", master_df['Target_Risk'].value_counts(normalize=True))

plt.figure(figsize=(10, 4))
sns.countplot(data=master_df, y='Primary_Condition', order=master_df['Primary_Condition'].value_counts().index, palette='Blues_r')
plt.title("Longitudinal Patient Cohort: Primary Condition Breakdown")
plt.xlabel("Patient Count")
plt.show()
```

### 4. Longitudinal Trajectory Trajectories (BP, Glucose & eGFR)


```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

sns.histplot(data=master_df, x='Delta_Systolic_BP', hue='Target_Risk', kde=True, ax=axes[0], palette='coolwarm')
axes[0].set_title("Longitudinal Systolic BP Delta (mmHg)")

sns.histplot(data=master_df, x='Delta_Glucose', hue='Target_Risk', kde=True, ax=axes[1], palette='coolwarm')
axes[1].set_title("Longitudinal Glucose Delta (mg/dL)")

sns.histplot(data=master_df, x='Delta_eGFR', hue='Target_Risk', kde=True, ax=axes[2], palette='coolwarm')
axes[2].set_title("Longitudinal eGFR Delta (mL/min)")

plt.tight_layout()
plt.show()
```

### 5. Acute Care Utilization & Readmission Impact


```python
plt.figure(figsize=(8, 4))
sns.boxplot(data=master_df, x='Target_Risk', y='Total_Healthcare_Cost', palette='Set2')
plt.title("Total Longitudinal Healthcare Cost by Risk Class")
plt.xlabel("Risk Category (0 = Low, 1 = High)")
plt.ylabel("Cost (USD)")
plt.show()
```
