### 1. Ingestion & NLP Setup


```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath('../src'))
from nlp_engine import ClinicalNLPEngine

notes_df = pd.read_csv('../data/clinical_notes.csv')
notes_df.head()
```

### 2. Clinical Note Aggregation & Acuity Scoring


```python
nlp_engine = ClinicalNLPEngine(max_features=40, model_dir='../models')
patient_notes = nlp_engine.aggregate_patient_notes(notes_df)
rule_features = nlp_engine.compute_rule_based_nlp_features(patient_notes)

print("Aggregated patient notes shape:", patient_notes.shape)
rule_features.head()
```

### 3. TF-IDF Clinical Vocabulary Analysis


```python
tfidf_df = nlp_engine.fit_transform_tfidf(patient_notes)
print("Top 15 TF-IDF Clinical Terms:")
print([col.replace('tfidf_', '') for col in tfidf_df.columns if col != 'Patient_ID'][:15])
```

### 4. Correlation of Clinical Acuity Score with Decompensation


```python
master_df = pd.read_csv('../data/processed_patient_master.csv')

plt.figure(figsize=(8, 4))
sns.kdeplot(data=master_df, x='NLP_Clinical_Acuity_Score', hue='Target_Risk', common_norm=False, fill=True, palette='magma')
plt.title("Distribution of NLP Clinical Acuity Score by Patient Risk Outcome")
plt.xlabel("Composite Clinical Acuity Score")
plt.show()
```
